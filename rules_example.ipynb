```python
import json
from datetime import datetime
from causal_graphviz.plot_conditions import draw_causal_graph

RULES_FILE = "mutiply_call_prompt/rules.json"  # 你的 rules.json 文件路径
with open(RULES_FILE, "r", encoding="utf-8") as f:
    rules_data = json.load(f)

# 2️⃣ 根据 Hazard 名称提取对应的 Scenarios
target_hazard = "VCE_liquid_continuous"

# 找到匹配项
target_rule = next((item for item in rules_data if item["Hazard"] == target_hazard), None)

target_rule
```

```python
CONDITIONS_FILE = "mutiply_call_prompt\conditions.json"
```

```python
timestamp = datetime.now().isoformat(timespec="seconds").replace(":", "-")
graph_png = draw_causal_graph(
    target_rule["Scenarios"],
    rules=CONDITIONS_FILE,
    save_path=f"./runs/artifacts/example_graph_{timestamp}.png"
)

print("Graph generated:", graph_png)
```
